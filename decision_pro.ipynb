```python
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("zeyadmohamed26/netflix-customer-churn-and-engagement-analytics")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(f'{path}/netflix_customer_churn.csv')
df.head()
```

```python
df.columns
```

```python
df = df.drop(columns=['customer_id'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df.head()
```

```python
obj = LabelEncoder()

df['gender'] = obj.fit_transform(df['gender'])
df['subscription_type'] = obj.fit_transform(df['subscription_type'])
df['region'] = obj.fit_transform(df['region'])
df['device'] = obj.fit_transform(df['device'])
df['payment_method'] = obj.fit_transform(df['payment_method'])
df['favorite_genre'] = obj.fit_transform(df['favorite_genre'])
```

```python
df.head()
```

```python
df.columns
```

```python
x = df[['age', 'gender', 'subscription_type', 'watch_hours', 'last_login_days',
       'region', 'device', 'monthly_fee','payment_method',
       'number_of_profiles', 'avg_watch_time_per_day', 'favorite_genre']]

y = df['churned']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model = DecisionTreeClassifier()
model.fit(x_train,y_train)
```

```python
predict = model.predict(x_test)
print(predict)
```

```python
score = classification_report(y_test,predict)
print(score)
```

```python
import pickle

pickle.dump(model, open('decision_pro.pkl','wb'))
print("PKL File Created Successfully")
```
